```python
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import mnist

# Memuat data MNIST
(train_images, _), (_, _) = mnist.load_data()

# Memilih satu gambar dari dataset
sample_image = train_images[0]
```

```python
# Menampilkan gambar
plt.imshow(sample_image, cmap='gray')
plt.colorbar()  # Menambahkan colorbar untuk referensi intensitas
plt.title("Sample Image from MNIST Dataset")
plt.show()
```

```python
print("Size of the image:", sample_image.shape)
```

```python
# Membuat filter konvolusi
filter_3x3 = np.random.randn(3, 3)
```

```python
def convolution(image, kernel):
    # Mendapatkan ukuran gambar dan filter
    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape

    # Menghitung ukuran output
    output_height = image_height - kernel_height + 1
    output_width = image_width - kernel_width + 1

    # Membuat array kosong untuk menyimpan hasil konvolusi
    conv_result = np.zeros((output_height, output_width))

    # Melakukan konvolusi
    for y in range(output_height):
        for x in range(output_width):
            conv_result[y, x] = np.sum(image[y:y+kernel_height, x:x+kernel_width] * kernel)

    return conv_result

# Melakukan konvolusi pada gambar sampel dengan filter 3x3
convolved_image = convolution(sample_image, filter_3x3)

# Menampilkan gambar asli dan hasil konvolusi
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(sample_image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('Convolved Image')
plt.imshow(convolved_image, cmap='gray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Fungsi konvolusi
def convolution(image, kernel):
    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape
    output_height = image_height - kernel_height + 1
    output_width = image_width - kernel_width + 1
    conv_result = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            conv_result[y, x] = np.sum(image[y:y+kernel_height, x:x+kernel_width] * kernel)
    return conv_result

# Mengambil sample gambar
sample_image = train_images[0]

# Membuat filter konvolusi pertama dan kedua
filter_3x3_1 = np.random.randn(3, 3)
filter_3x3_2 = np.random.randn(3, 3)

# Melakukan konvolusi dengan filter pertama
convolved_image_1 = convolution(sample_image, filter_3x3_1)

# Melakukan konvolusi dengan filter kedua
convolved_image_2 = convolution(sample_image, filter_3x3_2)

# Menampilkan gambar hasil konvolusi
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.imshow(sample_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(convolved_image_1, cmap='gray')
plt.title('Convolved Image 1')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(convolved_image_2, cmap='gray')
plt.title('Convolved Image 2')
plt.axis('off')

plt.show()
```

```python
def max_pooling(image, pool_size, stride):
    pool_height, pool_width = pool_size
    output_height = (image.shape[0] - pool_height) // stride + 1
    output_width = (image.shape[1] - pool_width) // stride + 1
    pooled_image = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            pooled_image[y, x] = np.max(image[y*stride:y*stride+pool_height, x*stride:x*stride+pool_width])
    return pooled_image

# Menggunakan max pooling pada gambar yang telah dikonvolusi
pool_size = (2, 2)  # Ukuran pool 2x2
stride = 2  # Stride sebesar 2

pooled_image_1 = max_pooling(convolved_image_1, pool_size, stride)
pooled_image_2 = max_pooling(convolved_image_2, pool_size, stride)

# Menampilkan hasil pooling
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(pooled_image_1, cmap='gray')
plt.title('Pooled Image 1')
plt.subplot(1, 2, 2)
plt.imshow(pooled_image_2, cmap='gray')
plt.title('Pooled Image 2')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import mnist

# Fungsi konvolusi
def convolution(image, kernel):
    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape
    output_height = image_height - kernel_height + 1
    output_width = image_width - kernel_width + 1
    conv_result = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            conv_result[y, x] = np.sum(image[y:y+kernel_height, x:x+kernel_width] * kernel)
    return conv_result

# Fungsi max pooling
def max_pooling(image, size):
    image_height, image_width = image.shape
    pool_height, pool_width = size
    output_height = image_height // pool_height
    output_width = image_width // pool_width
    pooled_image = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            pooled_image[y, x] = np.max(image[y*pool_height:(y+1)*pool_height, x*pool_width:(x+1)*pool_width])
    return pooled_image

# Fungsi untuk flattening
def flatten(image):
    return image.flatten()

# Fungsi aktivasi relu
def relu(x):
    return np.maximum(0, x)

# Fungsi aktivasi softmax
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # untuk menghindari overflow
    return exp_x / np.sum(exp_x)

# Memuat data MNIST
(train_images, _), (_, _) = mnist.load_data()

# Memilih satu gambar dari dataset
sample_image = train_images[0]

# Membuat filter konvolusi
filter_3x3_1 = np.random.randn(3, 3)
filter_3x3_2 = np.random.randn(3, 3)

# Melakukan konvolusi dengan filter pertama
convolved_image_1 = convolution(sample_image, filter_3x3_1)
convolved_image_1 = relu(convolved_image_1)  # Aktivasi ReLU

# Melakukan konvolusi dengan filter kedua
convolved_image_2 = convolution(sample_image, filter_3x3_2)
convolved_image_2 = relu(convolved_image_2)  # Aktivasi ReLU

# Melakukan max pooling
pooled_image_1 = max_pooling(convolved_image_1, (2, 2))
pooled_image_2 = max_pooling(convolved_image_2, (2, 2))

# Flatten hasil pooling
flattened_image_1 = flatten(pooled_image_1)
flattened_image_2 = flatten(pooled_image_2)

# Membangun model neural network sederhana
# Fungsi aktivasi di setiap lapisan di sini digunakan hanya sebagai contoh
# Pada praktiknya, lapisan-lapisan tersebut biasanya sudah diatur secara otomatis
# oleh pustaka seperti Keras atau TensorFlow.

# Inisialisasi bobot dan bias
weights_1 = np.random.randn(flattened_image_1.shape[0], 128)  # Bobot untuk fully connected layer 1
weights_2 = np.random.randn(128, 10)                          # Bobot untuk fully connected layer 2
biases_1 = np.random.randn(128)                               # Bias untuk fully connected layer 1
biases_2 = np.random.randn(10)                                # Bias untuk fully connected layer 2

# Lapisan fully connected layer
def fully_connected(x, weights, biases):
    return np.dot(x, weights) + biases

# Forward pass
fc_layer_1 = fully_connected(flattened_image_1, weights_1, biases_1)
fc_layer_1 = relu(fc_layer_1)  # Aktivasi ReLU
fc_layer_2 = fully_connected(fc_layer_1, weights_2, biases_2)

# Menggunakan softmax sebagai output layer
output = softmax(fc_layer_2)

print("Output Image 1 (Softmax):", output)
#jika ingin menampilkan gambar secara keseluruhan adalah sebagai berikut,

import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import mnist

# Fungsi konvolusi
def convolution(image, kernel):
    image_height, image_width = image.shape
    kernel_height, kernel_width = kernel.shape
    output_height = image_height - kernel_height + 1
    output_width = image_width - kernel_width + 1
    conv_result = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            conv_result[y, x] = np.sum(image[y:y+kernel_height, x:x+kernel_width] * kernel)
    return conv_result

# Fungsi max pooling
def max_pooling(image, size):
    image_height, image_width = image.shape
    pool_height, pool_width = size
    output_height = image_height // pool_height
    output_width = image_width // pool_width
    pooled_image = np.zeros((output_height, output_width))
    for y in range(output_height):
        for x in range(output_width):
            pooled_image[y, x] = np.max(image[y*pool_height:(y+1)*pool_height, x*pool_width:(x+1)*pool_width])
    return pooled_image

# Fungsi untuk flattening
def flatten(image):
    return image.flatten()

# Fungsi aktivasi relu
def relu(x):
    return np.maximum(0, x)

# Fungsi aktivasi softmax
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # untuk menghindari overflow
    return exp_x / np.sum(exp_x)

# Memuat data MNIST
(train_images, _), (_, _) = mnist.load_data()

# Memilih satu gambar dari dataset
sample_image = train_images[0]

# Membuat filter konvolusi
filter_3x3_1 = np.random.randn(3, 3)
filter_3x3_2 = np.random.randn(3, 3)

# Melakukan konvolusi dengan filter pertama
convolved_image_1 = convolution(sample_image, filter_3x3_1)
convolved_image_1 = relu(convolved_image_1)  # Aktivasi ReLU

# Melakukan konvolusi dengan filter kedua
convolved_image_2 = convolution(sample_image, filter_3x3_2)
convolved_image_2 = relu(convolved_image_2)  # Aktivasi ReLU

# Melakukan max pooling
pooled_image_1 = max_pooling(convolved_image_1, (2, 2))
pooled_image_2 = max_pooling(convolved_image_2, (2, 2))

# Flatten hasil pooling
flattened_image_1 = flatten(pooled_image_1)
flattened_image_2 = flatten(pooled_image_2)

# Membangun model neural network sederhana
# Fungsi aktivasi di setiap lapisan di sini digunakan hanya sebagai contoh
# Pada praktiknya, lapisan-lapisan tersebut biasanya sudah diatur secara otomatis
# oleh pustaka seperti Keras atau TensorFlow.

# Inisialisasi bobot dan bias
weights_1 = np.random.randn(flattened_image_1.shape[0], 128)  # Bobot untuk fully connected layer 1
weights_2 = np.random.randn(128, 10)                          # Bobot untuk fully connected layer 2
biases_1 = np.random.randn(128)                               # Bias untuk fully connected layer 1
biases_2 = np.random.randn(10)                                # Bias untuk fully connected layer 2

# Lapisan fully connected layer
def fully_connected(x, weights, biases):
    return np.dot(x, weights) + biases

# Forward pass
fc_layer_1 = fully_connected(flattened_image_1, weights_1, biases_1)
fc_layer_1 = relu(fc_layer_1)  # Aktivasi ReLU
fc_layer_2 = fully_connected(fc_layer_1, weights_2, biases_2)

# Menggunakan softmax sebagai output layer
output = softmax(fc_layer_2)

print("Output Image 1 (Softmax):", output)

# Menampilkan gambar hasil konvolusi dan pooling
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1)
plt.imshow(sample_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(convolved_image_1, cmap='gray')
plt.title('Convolved Image 1')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(pooled_image_1, cmap='gray')
plt.title('Pooled Image 1')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(convolved_image_2, cmap='gray')
plt.title('Convolved Image 2')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(pooled_image_2, cmap='gray')
plt.title('Pooled Image 2')
plt.axis('off')

# Menampilkan vektor hasil pooling
plt.subplot(2, 3, 6)
plt.bar(np.arange(len(flattened_image_1)), flattened_image_1)
plt.title('Flattened Image 1')
plt.xlabel('Pixel Index')
plt.ylabel('Pixel Value')

plt.tight_layout()
plt.show()

# Menampilkan vektor hasil fully connected layer
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(np.arange(len(fc_layer_1)), fc_layer_1)
plt.title('Fully Connected Layer 1')
plt.xlabel('Neuron Index')
plt.ylabel('Activation')

plt.subplot(1, 2, 2)
plt.bar(np.arange(len(output)), output)
plt.title('Softmax Output')
plt.xlabel('Class Index')
plt.ylabel('Probability')

plt.tight_layout()
plt.show()
```
